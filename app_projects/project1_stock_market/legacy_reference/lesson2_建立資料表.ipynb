```python
#建立資料表
import psycopg2
import os
conn = psycopg2.connect(host="localhost",database='stock_market',user=os.environ["POSTGRES_USER"],password=os.environ["POSTGRES_PASSWORD"])

#參考文件
#https://www.psycopg.org/docs/usage.html#transactions-control

with conn: #如果沒有raise exception 就自動執行commit, 有就自動執行rollback
    with conn.cursor() as cursor: #自動關閉cursor
        #建立市場資料表
        sql = '''
            CREATE TABLE IF NOT EXISTS 市場(
            name VARCHAR(20),
            country VARCHAR(20),
            PRIMARY KEY(name)
            );
        '''
        cursor.execute(sql)
    
    with conn.cursor() as cursor:#自動關閉cursor
        #建立股市資料表
        #NUMERIC(17,10):整數最多 7 位、小數 10 位;指數有 5 位數,不能用 NUMERIC(7,6)(最大只能到 9.999999)
        sql = '''
            CREATE TABLE IF NOT EXISTS 股市(
            stock_id SERIAL,
            date DATE,
            open NUMERIC(17,10),
            high NUMERIC(17,10),
            low NUMERIC(17,10),
            close NUMERIC(17,10),
            adj_close NUMERIC(17,10),
            volume BIGINT DEFAULT 0,
            name VARCHAR(20),
            PRIMARY KEY(stock_id),
            FOREIGN KEY(name) REFERENCES 市場(name) 
            ON DELETE NO ACTION
            ON UPDATE CASCADE
            );
        '''
        cursor.execute(sql) 

conn.close()
```
