```python
#台灣大盤[^TWII]
#auto_adjust=False 才會有 Adj Close(還原權值收盤價)欄位
taiwan = yf.Ticker('^TWII').history(period='max', auto_adjust=False)
taiwan
```

```python
from datetime import datetime
#使用index比較符合我們的需求
all_data = taiwan.to_dict(orient='index')
#先取出key,再一筆一筆取出資料(只印出前 3 筆)
for key,value in list(all_data.items())[:3]:
    print(key)
    print(type(key))
    print(value)
    print("=========")
```
